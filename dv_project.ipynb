```python
match_wins = df['match_winner']
match_wins = match_wins.value_counts()
sns.barplot(y = match_wins.index, x=match_wins.values)
```

```python
df1 = df[df['match_winner'] == df['toss_winner']]
count = df1['match_id'].count()
percent = (count*100)/len(df)
print(percent.round(2))
```

### Key Player Performances
1 Most "Player of the Match" Awards


2 Top Scorers


10 Best Bowling Figures[link text](https://)

```python
player_of_match = df['player_of_the_match'].value_counts().head(10)
sns.barplot(y = player_of_match.index, x = player_of_match.values, palette='rainbow')
```

```python
top_scorer = df.groupby(df['top_scorer'])['highscore'].sum()
top_scorer = top_scorer.sort_values(ascending = False).head(2)
sns.barplot(x = top_scorer.index, y = top_scorer.values, palette='rainbow')
```

```python
df.drop('highest_wickets',axis = 1 , inplace=True)
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x: x.split("--")[0]).astype(int)
best_bolwers = df.groupby(df['best_bowling'])['highest_wickets'].sum().sort_values(ascending = False).head(10)
sns.barplot(y = best_bolwers.index, x = best_bolwers.values, palette='rainbow')

```
